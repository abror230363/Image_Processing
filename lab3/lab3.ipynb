```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
image = cv2.imread("/Users/230363/Desktop/image.jpg.avif")

print(image.shape)
translation_matrix = np.float32([[1, 0, 100], [0, 1, 50]])

translated_image = cv2.warpAffine(image, translation_matrix, (image.shape[1], image.shape[0]))
print(translated_image.shape)
plt.imshow(cv2.cvtColor(translated_image, cv2.COLOR_BGR2RGB))
plt.title('Translated Image')
plt.axis('off')
plt.show()
```

```python
translated_image = translate_image(image, 0, -100)

plt.imshow(cv2.cvtColor(translated_image, cv2.COLOR_BGR2RGB))
plt.title("Shift: 100 Up")
plt.axis('off')
plt.show()
```

```python
(h, w) = image.shape[:2]
center = (w // 2, h // 2)
rotation_matrix = cv2.getRotationMatrix2D(center,
45, 1.0)
rotated_image = cv2.warpAffine (image,
rotation_matrix, (w, h))
plt.imshow(cv2.cvtColor (rotated_image,
cv2.COLOR_BGR2RGB))
plt.title('Rotated Image (45 degrees) ')
plt.axis('off')
plt.show()
```

```python
(h, w) = image.shape[:2]
center = (w // 2, h // 2)
rotation_matrix = cv2.getRotationMatrix2D(center,
90, 1.0)
rotated_image = cv2.warpAffine (image,
rotation_matrix, (w, h))
plt.imshow(cv2.cvtColor (rotated_image,
cv2.COLOR_BGR2RGB))
plt.title('Rotated Image (90 degrees) ')
plt.axis('off')
plt.show()
```

```python
(h, w) = image.shape[:2]
center = (w // 2, h // 2)
rotation_matrix = cv2.getRotationMatrix2D(center,
180, 1.0)
rotated_image = cv2.warpAffine (image,
rotation_matrix, (w, h))
plt.imshow(cv2.cvtColor (rotated_image,
cv2.COLOR_BGR2RGB))
plt.title('Rotated Image (180 degrees) ')
plt.axis('off')
plt.show()
```

```python
Factor = 2
scaling_matrix = cv2.getRotationMatrix2D ( (0, 0), 0, Factor)

scaled_image = cv2.warpAffine (image, scaling_matrix, (w*Factor, h*Factor))

plt.imshow(cv2.cvtColor(scaled_image, cv2.COLOR_BGR2RGB))
plt.title('Scaled Image (2x)')
plt.axis('off')
plt.show()
print (scaled_image. shape)
```

```python
factor = 0.5

scaled_half = cv2.resize(
    image,
    (int(w * factor), int(h * factor)),
    interpolation=cv2.INTER_AREA   
)

plt.imshow(cv2.cvtColor(scaled_half, cv2.COLOR_BGR2RGB))
plt.title("Scaled Image (0.5x)")
plt.axis('off')
plt.show()

print("Scaled Shape (0.5x):", scaled_half.shape)
```

```python
m = 0.67

shearing_matrix = np.float32([[1, m, 0], [0, 1, 0]])

sheared_image = cv2. warpAffine (image, shearing_matrix, (w, h))

plt.imshow(cv2.cvtColor (sheared_image, cv2.COLOR_BGR2RGB))
plt.title('Sheared Image (x-axis, factor ' + str (m)+') ')
plt.axis('off')
plt.show()
```

```python
factors = [0.3, 0.8, 1.2, -0.5]

for m in factors:
    shear_matrix = np.float32([
        [1, m, 0],
        [0, 1, 0]
    ])
    
    sheared = cv2.warpAffine(image, shear_matrix, (w + int(abs(m)*h), h))
    
    plt.figure(figsize=(6,6))
    plt.imshow(cv2.cvtColor(sheared, cv2.COLOR_BGR2RGB))
    plt.title(f"Shear X-axis (m = {m})")
    plt.axis('off')
    plt.show()
```

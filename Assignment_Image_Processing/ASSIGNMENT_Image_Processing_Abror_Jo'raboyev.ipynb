```python
import matplotlib.image as mpimg
import matplotlib.pyplot as plt



img_mp = mpimg.imread("/Users/230363/Documents/image.jpeg")   

print("Shape:", img_mp.shape)
print("Dtype:", img_mp.dtype)

plt.imshow(img_mp)
plt.axis("off")
plt.title("Matplotlib imread")
plt.show()

```

```python
from skimage import io, img_as_float
import matplotlib.pyplot as plt



img_sk = io.imread("/Users/230363/Documents/image.jpeg")     
img_sk_f = img_as_float(img_sk)                

print("Original dtype:", img_sk.dtype)
print("Float dtype:", img_sk_f.dtype)
print("Min pixel value:", img_sk_f.min())
print("Max pixel value:", img_sk_f.max())

plt.imshow(img_sk_f)
plt.axis("off")
plt.title("scikit-image img_as_float")
plt.show()
```

```python
import cv2
import matplotlib.pyplot as plt



img_cv_gray = cv2.imread("/Users/230363/Documents/image.jpeg", cv2.IMREAD_GRAYSCALE)  
img_cv_bgr  = cv2.imread("/Users/230363/Documents/image.jpeg", cv2.IMREAD_COLOR)      

img_cv_rgb = cv2.cvtColor(img_cv_bgr, cv2.COLOR_BGR2RGB)                

plt.imshow(img_cv_rgb)
plt.axis("off")
plt.title("OpenCV (converted to RGB)")
plt.show()
```

```python


edges = cv2.Canny(img_cv_gray, 100, 200)   #
plt.imshow(edges, cmap="gray")
plt.axis("off")
plt.title("Canny Edges")
plt.show()
```

```python
import glob
import cv2
import matplotlib.pyplot as plt


path = "/Users/230363/Documents/image.jpeg"

for file in glob.glob(path):
    img_bgr = cv2.imread(file)
    if img_bgr is None:
        continue  

    img_rgb = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2RGB)

    plt.imshow(img_rgb)
    plt.axis("off")
    plt.title(file)
    plt.show()
```
